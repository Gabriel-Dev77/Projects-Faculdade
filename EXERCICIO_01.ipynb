```python
from google.colab import files
from google.colab.patches import cv2_imshow
import cv2
import numpy as np

uploaded = files.upload()
nome_arquivo = list(uploaded.keys())[0]

imagem = cv2.imread(nome_arquivo)

print("Imagem Original")
cv2_imshow(imagem)

B, G, R = cv2.split(imagem)

print("Canal Vermelho - Intensidade")
cv2_imshow(R)

print("Canal Verde - Intensidade")
cv2_imshow(G)

print("Canal Azul - Intensidade")
cv2_imshow(B)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
from google.colab.patches import cv2_imshow
import cv2

imagem = cv2.imread(nome_arquivo)

hsv = cv2.cvtColor(imagem, cv2.COLOR_BGR2HSV)

H, S, V = cv2.split(hsv)

print("Canal H - Tonalidade (Hue)")
cv2_imshow(H)

print("Canal S - Saturação")
cv2_imshow(S)

print("Canal V - Valor/Brilho")
cv2_imshow(V)

# Canal H: indica a cor (tonalidade) de cada pixel. Áreas claras mostram.

# Canal S: indica o quão viva/pura é a cor. Áreas claras = cores fortes
# e saturadas; áreas escuras = cores fracas, esbranquiçadas ou acinzentadas.

# Canal V: indica o brilho de cada pixel. Áreas claras = regiões mais
# iluminadas da imagem original; áreas escuras = regiões mais sombreadas.

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
from google.colab import files
from google.colab.patches import cv2_imshow
import cv2
import numpy as np

uploaded = files.upload()
nome_arquivo_vermelho = list(uploaded.keys())[0]
imagem_vermelha = cv2.imread(nome_arquivo_vermelho)

B, G, R = cv2.split(imagem_vermelha)

lado_a_lado = np.hstack((cv2.cvtColor(imagem_vermelha, cv2.COLOR_BGR2RGB), cv2.cvtColor(R, cv2.COLOR_GRAY2RGB)))

print("Imagem Original  x  Canal R")
cv2_imshow(cv2.cvtColor(lado_a_lado, cv2.COLOR_RGB2BGR))

# O objeto vermelho fica mais claro no canal R porque nesses pixels o
# valor de R é alto, enquanto nos canais G e B é baixo. Isso ajuda em
# tarefas como detecção de objetos e segmentação por cor.

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
from google.colab import files
from google.colab.patches import cv2_imshow
import cv2

uploaded = files.upload()
nome_arquivo_sombra = list(uploaded.keys())[0]

imagem_sombra = cv2.imread(nome_arquivo_sombra)

hsv_sombra = cv2.cvtColor(imagem_sombra, cv2.COLOR_BGR2HSV)

H, S, V = cv2.split(hsv_sombra)

print("Canal S - Saturação")
cv2_imshow(S)

print("Canal V - Valor/Brilho")
cv2_imshow(V)

# O canal V cai nas sombras, pois representa o brilho da imagem.
# O canal S também pode cair em sombras profundas, pois com pouca
# luz fica mais difícil identificar a cor real do pixel.

cv2.waitKey(0)
cv2.destroyAllWindows()
```
